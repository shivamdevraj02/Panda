# Pandas Setup in VS Code

## 1. Python Check

VS Code Terminal me check karo:

```bash
python --version
```

Recommended:

```text
Python 3.11.0
```

---

## 2. Project Folder

Example:

```text
E:\AI-ML\Pandda
```

VS Code me isi folder ko open karo.

---

## 3. Virtual Environment Create

Terminal me:

```bash
python -m venv .venv
```

Isse project ke andar `.venv` folder banega.

---

## 4. Virtual Environment Activate

Windows:

```bash
.venv\Scripts\activate
```

Successful activation ke baad terminal me:

```text
(.venv) E:\AI-ML\Pandda>
```

dikhega.

> Agar `(.venv)` already dikh raha hai, environment active hai. Dobara activate ya create karne ki zarurat nahi.

---

## 5. Pandas Install

Sirf Pandas ke liye:

```bash
python -m pip install pandas
```

Check:

```bash
python -c "import pandas as pd; print(pd.__version__)"
```

Agar version print ho gaya → Pandas successfully installed.

---

## 6. VS Code Python Interpreter Select

`Ctrl + Shift + P`

→ **Python: Select Interpreter**

→ select:

```text
.venv (3.11.0)
Python 3.11.0
.venv\Scripts\python.exe
```
