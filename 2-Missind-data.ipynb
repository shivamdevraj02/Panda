## Finding missing Data 

```python
import numpy as np ;
import pandas as pd;
```

```python
data = {
    'A': [1, 2, 3, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}

df = pd.DataFrame(data)

df
```

```python
df.isna() 
```

```python
df.isna().sum()
```

```python
# checking for any null value is availabe

df.isna().any()
```

Removing the missing data 

```python
df
```

```python
# rows wise drop hoga
df.dropna()
```

```python

# thresh =3 means Jis row mein kam se kam 3 non-null (non-NaN) values hain, us row ko rakho.

df.dropna(thresh=3)
```

 Filling the missing datam

```python
df
```

```python
# ye sab jitne bhi changes ho rhe he wo real table me nii ho rha he uske liye inplace = True likhna padega.

df.fillna(0)
```

```python
# agr hame alg alg coloumn me different values dena ho to :--
#colounms wise hoga

values ={'A':0,'B':100,'C':300,'D':400}
df.fillna(value=values)
```

```python
# ab yaha par mean se data fill hoga
df.fillna(df.mean());
df
```
