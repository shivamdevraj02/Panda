```python
import numpy as np;
import pandas as pd;
```

```python
employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

# DataFrame 2: Salary information
salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})
```

```python
employees
```

```python
salaries
```

Now we going to merge these two tables on the basis of different ways :--

```python
pd.merge(employees ,salaries)
```

```python
# here employee_id is showing on which basing these two table should be merged

pd.merge(employees,salaries, on="employee_id")
```

```python
#  Dono DataFrames mein jo matching rows/common values hain, sirf unko rakho.

# Pandas yahan automatically common column employee_id ko matching ke liye use karega. 

pd.merge(employees ,salaries , how="inner")
```

```python
# how="outer" dono DataFrames ke saare rows ko include karta hai.
# Jahan matching data nahi hota, wahan NaN aa jata hai.

pd.merge(employees,salaries,how="outer")
```

```python
# how = "left" means yaha par (employees and salaries) me se employee ko priority milege kyunki wo left me he .

pd.merge(employees,salaries,how="left")
```

```python
# how = "right" means yaha par (employees and salaries) me se salaries ko priority milege kyunki wo right me he .

pd.merge(employees,salaries,how="right")
```

### Joining two dataFrames

```python
df1 = pd.DataFrame({
    'name:': ['Alice' ,'bob','charlie']
}, index= [1,2,3])

df2 = pd.DataFrame({
    'Score' : [85,90,75]
},index=[2,3,4])
```

```python
df1
```

```python
df2
```

```python
df1.join(df2)
```

```python
df1.join(df2,how='outer')
```
